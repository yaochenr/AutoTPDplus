```python
from rdkit import Chem

smiles = "O=C(N1CCN(CCC#CC2=CC=CC3=C2CN(C4CCC(NC4=O)=O)C3=O)CC1)N5CCN(CCN6CCN(C7=CC(NC8=NC=C(C(NC9=C(C)C=CC=C9Cl)=O)S8)=NC(C)=N7)CC6)CC5"

mol = Chem.MolFromSmiles(smiles)

inchi = Chem.MolToInchi(mol)
inchikey = Chem.InchiToInchiKey(inchi)

print("Standard InChI:", inchi)
print("Standard InChIKey:", inchikey)
```
